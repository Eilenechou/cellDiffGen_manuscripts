```python
import numpy as np
from matplotlib import font_manager
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.legend_handler import HandlerTuple
font_manager.fontManager.addfont("../../../../arial.ttf")
plt.rcParams["font.family"] = "Arial"
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
    "pdf.fonttype": 42,
})
plt.rcParams["hatch.linewidth"] = 1

heights = np.array([0.990,-0.007,0.979,0.004,0.955,0.005,0.462,-0.004,0.948,0.002,0.826,-0.007,0.592,0.002]).flatten()
x = np.arange(len(heights))
x = x + x // 2  # Gap between pairs

fig, ax = plt.subplots(figsize=(6, 3), dpi=200)

# bars = ax.bar(x, heights, color=["#FEDCCA", "#F58A43"] * 15,
#               width=0.63, edgecolor="black", linewidth=0.5)

# colors = ["#6666CC", "#3399CC"] * 5

colors = ["#6666CC", "#3399CC"] * (len(heights) // 2)

# Layer 1: translucent fill
bars = ax.bar(
    x, heights, bottom=0,
    color=[c + "40" for c in colors],
    edgecolor="none",
    width=0.63, linewidth=1,
)

# Layer 2: hatch and opaque edges
hatch_bars = ax.bar(
    x, heights, bottom=0,
    facecolor="none",
    edgecolor=colors,
    width=0.63, linewidth=1,
    hatch="xxx",
)
labels = ax.bar_label(bars, labels=[f"{v:.3f}" for v in heights], padding=4, fontsize=10,annotation_clip=False)
for label, value, bar in zip(labels, heights, bars):
    # if value >= 0.3:
    #     # print(value)
    #     label.xy = (bar.get_x() + bar.get_width() / 2, 1.01)
    #     # print(label.xy)
    #     label.set_position((0, 0))
    if value >=0:
        a, b = label.get_position()
        label.set_position((a + 3, b-2))
    else:
        a, b = label.get_position()
        label.set_position((a, b))
ax.spines["top"].set_visible(False)
ax.axhline(y=0, color="gray", linestyle=":", linewidth=1)
# ax.axhline(y=0.5, color="gray", linestyle=":", linewidth=1)
# ax.axhline(y=1, color="gray", linestyle=":", linewidth=2)

for position in range(2, 18, 3):
    ax.axvline(position, color="gray", linestyle=":", linewidth=1)

ax.set_xticks((x[::2] + x[1::2]) / 2)
ax.set_xticklabels(
    [f"Cluster {i}\n(n={j})" for i, j in zip(range(7), [406, 356, 288, 60,140,89,35])],
    rotation=45,
    ha="center"
)
ax.set_ylim(-0.1, 1)
ax.set_yticks([-0.1, 0, 0.5,1])
ax.set_ylabel("Pearson r")
# ax.set_title("POU5F1B mRNA", loc="left", color="#ED663B")
ax.legend(frameon=False, loc="center", bbox_to_anchor=(1.02, 0.5))

# plt.tight_layout()
fig.savefig("bar_plot_part3.pdf", bbox_inches="tight")
plt.show()
```
