```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({"font.family": "Arial", "font.size": 11})

x = np.arange(5)

series = {
    "Series 1": dict(y=[1, 2, 1.5, 3, 2.5], color="#6A0DAD", marker="v"),   # purple triangle-down
    "Series 2": dict(y=[2, 2.5, 2, 2.8, 3],   color="#444444", marker="D"), # dark diamond
    "Series 3": dict(y=[1.5, 1.8, 2.2, 2, 2.6], color="#222222", marker="^"), # black triangle-up
    "Series 4": dict(y=[0.5, 1, 1.2, 1.5, 1.8], color="#E07B1E", marker="o"), # orange circle
    "Series 5": dict(y=[3, 3.2, 3.1, 3.4, 3.6], color="#3355CC", marker="s"), # blue square
    "Series 6": dict(y=[2.5, 2.6, 2.4, 2.7, 2.9], color="#3355CC", marker="o", markerfacecolor="none"), # blue open circle
}

fig, ax = plt.subplots(figsize=(4, 3), dpi=200)

for label, s in series.items():
    ax.plot(
        x, s["y"],
        marker=s["marker"],
        color=s["color"],
        markerfacecolor=s.get("markerfacecolor", s["color"]),
        markeredgecolor=s["color"],
        linewidth=1.5,
        markersize=7,
        label=label,
    )

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.spines[["top", "right"]].set_visible(False)
ax.legend(frameon=False, fontsize=9, loc="upper left", bbox_to_anchor=(1.02, 1))

plt.tight_layout()
fig.savefig("line_plot.pdf", bbox_inches="tight")
plt.show()
```
