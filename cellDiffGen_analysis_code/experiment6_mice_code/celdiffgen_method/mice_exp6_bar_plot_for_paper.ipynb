```python
import numpy as np
import matplotlib.pyplot as plt

heights = np.array([0.937, 0.868, 0.918, 0.793, 0.982, 0.950])
x = np.arange(len(heights))
x = x + x // 2  # Gap between pairs

fig, ax = plt.subplots(figsize=(6, 2), dpi=200)

# Both bars in each pair have the same color and hatch
colors = ["#32489A", "#E22831"] * 3
hatches = ["", "", "/", "/", "///", "///"]

# Layer 1: translucent fill
bars = ax.bar(
    x, heights,
    color=[c + "90" for c in colors],
    edgecolor="none",
    width=0.5,
)

# Layer 2: opaque outlines and hatches
hatch_bars = ax.bar(
    x, heights,
    facecolor="none",
    edgecolor=colors,
    width=0.5,
    linewidth=1,
)

for bar, hatch in zip(hatch_bars, hatches):
    bar.set_hatch(hatch)

# Labels aligned above the bars
for i, (bar, value) in enumerate(zip(bars, heights)):
    if value in [0.937, 0.918, 0.982]:
            ax.text(
            bar.get_x() + bar.get_width() / 2, 1.02,
            f"Pearson r = {value:.3f}",
            ha="center", va="bottom", fontsize=8,
            clip_on=False
        )
    else:
        ax.text(
            bar.get_x() + bar.get_width(), 1.02,
            f"$R^2$ = {value:.3f}",
            ha="center", va="bottom", fontsize=8,
            clip_on=False
        )

ax.spines["top"].set_visible(False)

for y, linewidth in [(1, 2), (0.8, 1), (0.9, 1)]:
    ax.axhline(y, color="gray", linestyle=":", linewidth=linewidth)

for position in [2, 5]:
    ax.axvline(position, color="gray", linestyle=":", linewidth=1)

ax.set_xticks((x[::2] + x[1::2]) / 2)
ax.set_xticklabels(
    [
        "Train day 7 and day 10\nPredict day 19 from day 7",
        "Train day 7 and day 13\nPredict day 19 from day 7",
        "Train day 7 and day 18\nPredict day 19 from day 7",
    ],
    ha="center",
)

ax.set_ylim(0, 1)
ax.set_yticks([0, 0.8, 0.9, 1])

plt.tight_layout()
fig.savefig("bar_plot_mice_figureA.pdf", bbox_inches="tight")
plt.show()
```
