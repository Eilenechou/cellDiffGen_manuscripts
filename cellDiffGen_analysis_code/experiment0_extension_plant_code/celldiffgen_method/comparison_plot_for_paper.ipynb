```python
import numpy as np
from matplotlib import font_manager
import matplotlib.pyplot as plt

font_manager.fontManager.addfont("../../../../arial.ttf")
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 15,
    "pdf.fonttype": 42,
})

pearsonr = np.array([0.519, 0.007,0.814,0.003,0.771,-0.002])

x = np.array([0,0.92, 2.63, 3.57, 5.08, 6])
# x = np.array([0, 1, 2.5, 3.5, 5, 6])
fill_colors = ["#8273B8", "#B1D4E7"] * 3
edge_colors = ["#4B3A80", "#B1D4E7"] * 3 
ylim = (-0.2, 1)

fig, ax = plt.subplots(figsize=(4, 4), dpi=200)

bars = ax.bar(x, pearsonr, color=fill_colors, edgecolor=edge_colors,
              linewidth=1.2, width=0.9, capsize=3, error_kw={"linewidth": 1.5})

# # make bars start from the bottom of ylim instead of 0
# bottom = ylim[0]
# for bar, m in zip(bars, pearsonr):
#     bar.set_y(bottom)
#     bar.set_height(m - bottom)

labels = ax.bar_label(bars, labels=[f"{v:.3f}" for v in [0.519, 0.007,0.814,0.003,0.771,-0.002]], padding=2, fontsize=13,annotation_clip=False)
for label, value, bar in zip(labels, pearsonr, bars):
    if value < 0:
        a, b = label.get_position()
        label.set_position((a + 3, b))
    else:
        a, b = label.get_position()
        label.set_position((a + 3, b-3))

ax.set_xticks((x[::2] + x[1::2]) / 2)
ax.set_xticklabels([f"Day {i}" for i in [0, 12, 21]])
ax.set_ylim(ylim)
ax.set_yticks([-0.2, 0, 0.5, 1])
ax.set_ylabel("Pearson r")
# ax.spines[["top", "right"]].set_visible(False)
ax.axhline(y=0, color="gray", linestyle=":", linewidth=1)
# ax.axhline(y=1, color="gray", linestyle=":", linewidth=2)
handles = [plt.Rectangle((0, 0), 1, 1, color=c) for c in fill_colors[:2]]
ax.legend(handles, ["cellDiffGen", "Squidiff"], frameon=False,
          loc="lower center", bbox_to_anchor=(0.5, 1.02), ncol=2)

fig.savefig("comparison_plot_pearsonr.pdf", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
from matplotlib import font_manager
import matplotlib.pyplot as plt

font_manager.fontManager.addfont("../../../../arial.ttf")
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 12,
    "pdf.fonttype": 42,
})

r2 = np.array([-1.053,-175151,0.631,-204033,-0.072,-198467])

x = np.array([0, 1, 2.5, 3.5, 5, 6])
fill_colors = ["#8273B8", "#B1D4E7"] * 3
edge_colors = ["#4B3A80", "#B1D4E7"] * 3 
ylim = (-0.2, 1)

fig, ax = plt.subplots(figsize=(4, 1.5), dpi=200)

bars = ax.bar(x, r2, color=fill_colors, edgecolor=edge_colors,
              linewidth=1.2, width=0.9, capsize=3, error_kw={"linewidth": 1.5})

# # make bars start from the bottom of ylim instead of 0
# bottom = ylim[0]
# for bar, m in zip(bars, r2):
#     bar.set_y(bottom)
#     bar.set_height(m - bottom)

label_texts = [f"{v:.3f}" if v >= 0 else "<0" for v in r2]
labels = ax.bar_label(bars, labels=label_texts, padding=2, fontsize=13)
for label, value, bar in zip(labels, r2, bars):
    if value < 0:
        print(value)
        label.xy = (bar.get_x() + bar.get_width() / 2, -0.03)
        # print(label.xy)
        label.set_position((0, 0))
    else:
        print(value)
        a, b = label.get_position()
        label.set_position((a + 3, b-3))
ax.set_xticks((x[::2] + x[1::2]) / 2)
ax.set_xticklabels([f"Day {i}" for i in [0, 12, 21]])
ax.set_ylim(ylim)
ax.set_yticks([0, 0.5, 1])
ax.set_ylabel(r"$R^2$")
# ax.spines[["top", "right"]].set_visible(False)
ax.axhline(y=0, color="gray", linestyle=":", linewidth=1)


fig.savefig("comparison_plot_r2.pdf", bbox_inches="tight")
plt.show()
```
